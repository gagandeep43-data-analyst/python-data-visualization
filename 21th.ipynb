```python
# generate daily dates in a given range
```

```python
pd.date_range(start='2023/1/5',end='2023/2/28',freq='2D')
```

```python
# business days
```

```python
pd.date_range(start='2023/1/5',end='2023/2/28',freq='B')
```

```python
# W-> one week per day
```

```python
pd.date_range(start='2023/1/5',end='2023/2/28',freq='W-THU')
```

```python
# H-> Hourly data
```

```python
pd.date_range(start='2023/1/5',end='2023/1/28',freq='6h')
```

```python
# M-> Month end
```

```python
pd.date_range(start='2023/1/5',end='2023/4/28',freq='MS')
```

```python
# MS->Month Start
```

```python
pd.date_range(start='2023/1/5',end='2023/2/28',freq='MS')
```

```python
# A-> Year end
```

```python
pd.date_range(start='2023/1/5',end='2023/2/28',freq='YE')
```

```python
# using periods (number of results)
```

```python
pd.date_range(start='2023/1/5',end='2025/5/28',periods=25,freq='ME')
```

```python
# to_datetime function
```

```python
# converts existing objects to pandas timestamp/datetimeindex
```

```python
s=pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
pd.to_datetime(s)
```

```python
# with errors
```

```python
s=pd.Series(['2023/1/1','2022/1/1','2021/13/1'])
pd.to_datetime(s,errors='coerce').dt.day_name()
```

```python
df=pd.read_csv("expense_data.csv")
df
```

```python
df.head()
```

```python
df['Date']=pd.to_datetime(df['Date'])
df.info()
```

```python
# dt accessor
```

```python
df['Date'].dt.is_quarter_start.sample(20)
```

```python
# Plot graph
```

```python
plt.plot(df['Date'],df['INR'])
plt.show()
```

```python
# day name wise bar chart
```

```python
df['day_name']=df['Date'].dt.day_name()
df.head()
```

```python
df.groupby('day_name')['INR'].mean().plot(kind='pie')
plt.show()
```

```python
# month name wise bar chart
```

```python
df['month_name']=df['Date'].dt.month_name()
df.head()
```

```python
df.groupby('month_name')['INR'].sum().plot(kind='bar')
plt.show()
```
