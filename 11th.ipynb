# Pandas GroupBy - Big Programe 1

```python
import pandas as pd
```

```python
df=pd.read_csv("imdb-top-1000.csv.xls")
```

```python
df
```

```python
df["Released_Year"]=pd.to_numeric(df["Released_Year"],errors="coerce")
df
```

```python
df["IMDB_Rating"]=pd.to_numeric(df["IMDB_Rating"],errors="coerce")
df
```

```python
df["No_of_Votes	"]=pd.to_numeric(df["No_of_Votes"],errors="coerce")
df
```

```python
df["Gross"]=pd.to_numeric(df["Gross"],errors="coerce")
df
```

```python
df["Metascore"]=pd.to_numeric(df["Metascore"],errors="coerce")
df
```

```python
df["Decade"]=(df["Released_Year"]//10)*10
df
```

```python
genre_count=df.groupby("Genre").size().sort_values(ascending=False)
genre_count
```

```python
genre_rating=df.groupby("Genre")["IMDB_Rating"].sum().sort_values(ascending=False)
genre_rating
```

```python
genre_analysis=df.groupby("Genre",as_index=False).agg(
    Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","max"),
    Highest_Rating=("IMDB_Rating","mean"),
    Average_Runtime=("Runtime","mean"),
    Total_Votes=("No_of_Votes","sum"),
    Total_Gross=("Gross","sum")
    )
genre_analysis
```

```python
genre_analysis=genre_analysis.sort_values("Average_Rating",ascending=False)
genre_analysis
```

```python
popular_genres=genre_analysis[genre_analysis["Movies"] >=20]
popular_genres
```

```python
high_rated_genre=genre_analysis[(genre_analysis["Movies"] >=10)&(genre_analysis["Average_Rating"] >=7.5)
    ]
high_rated_genre
```

```python
best_genre=genre_analysis.loc[genre_analysis["Average_Rating"].idxmax()]
best_genre
```

```python
most_movies_genre=genre_analysis.loc[genre_analysis["Movies"].idxmin()]
most_movies_genre
```

```python
highest_gross_genre=genre_analysis.loc[genre_analysis["Total_Gross"].idxmax()]
highest_gross_genre
```

```python
decade_analysis=df.groupby("Decade",as_index=False).agg(
     Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","max"),
    Highest_Rating=("IMDB_Rating","mean"),
    Average_Runtime=("Runtime","mean"),
    Total_Votes=("No_of_Votes","sum"),
    Total_Gross=("Gross","sum")
    )
decade_analysis
    
```

```python
decade_analysis=decade_analysis.sort_values("Decade")
decade_analysis
```

```python
top_decade=decade_analysis.sort_values("Average_Rating",ascending=False).head(5)
top_decade
```
