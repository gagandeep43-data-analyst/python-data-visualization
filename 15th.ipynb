**Q1. Students dataset me student_id aur name kouse karke student_info banao aur Month 1 registrations ke sath.**

**Answer:**

```python
student_info=students[["student_id","name"]]
result1=pd.merge(month1,student_info,
                 on="student_id",
                 how="left")
result1
                       
```

**Q2. Course dataset se sirf course_id aur course_namelaker Month2 registration ke saath merge karo.**

**Answer:**

```python
course_info=courses[["course_id","course_name"]]
result2=pd.merge(month2, course_info,on="course_id",how="left")
result2
```

```python
result3=month1.merge(
    students[["student_id","name"]],
              on="student_id",
              how="left"
).merge(
    courses[["course_id","course_name","price"]],
             on="course_id",
             how="left"
)
result3
```

**Q4. Month 1 aur Month 2 registrations ko students ke naam ke sath merge karke ek dataFrameme combine karo.**

**Answer:**

```python
month1_named=month1.merge(
    students,
    on="student_id",
    how="left"
)
month2_named=month2.merge(
    students,
    on="student_id",
    how="left"
)
dataframe = pd.concat([month1_named,month2_named],ignore_index=True)
dataframe
```

**Q5. Student dataset ko khud ke sath partner column ke through merge karke student aur partner dono ke name.**

**Answer:**

```python
students
```

```python
student_partner=pd.merge(
    students,
    students[["student_id","name"]],
    left_on="partner",
    right_on="student_id",
    how="left",
    suffixes=("_student","_partner")
)
student_partner[["student_id_student","name_student","partner","name_partner"]]
```

**Q6.month1 registrations ko courses ke aath merge karke price aur course ki averge price compare karo.**

**Answer:**

```python
month1
```

```python
courses
```

```python
price_data=month1.merge(
    courses[["course_id","course_name","price"]],
    on="course_id",
    how="left"
)
price_data
```

```python
price_data["course_average_price"]=price_data.groupby(
    "course_id"
)["price"].transform("mean")

price_data
```

```python
price_data["price_difference"]=(price_data["price"]-price_data["course_average_price"])
price_data[[
"course_id","course_name","price",
"course_average_price","price_difference"
]]
price_data
```

**Q7. Month 1 aur month 2 ke course registrations ko course_id ke basis par compare karo.**

**Answer:**

```python
month1
```

```python
month2
```

```python
m1_count=month1.groupby("course_id").size().reset_index(name="month1_count")
m2_count=month2.groupby("course_id").size().reset_index(name="month2_count")
print(m1_count)
print(m2_count)
```

```python
comparison=pd.merge(
    m1_count,
    m2_count,
    on="course_id",
    how="outer"
).fillna(0)
comparison
```

```python
comparison=comparison.merge(
    courses[["course_id","course_name"]],
    on="course_id",
    how="left"
)
comparison
```

```python
comparison["difference"]=(
   comparison["month2_count"]- comparison["month1_count"]
)
comparison
    
```

**Q8. un students ko nikalo jinki month2 registration month1 se jyada hai.**

**Answer:**

```python
comparison
```

```python
more_registrations=comparison[
    comparison["month2_count"]>comparison["month1_count"]
    ]
more_registrations
```

**Q9. students aur courses ko merge karne ke baad course_wise unique students ki count nikalo.**

**Answer:**

```python
students_courses=month1.merge(
    students[["student_id","name"]],
    on="student_id",
    how="left"
).merge(
    courses[["course_id","course_name"]],
    on="course_id",
how="left"
)
students_courses
```

```python
unique_students=students_courses.groupby(
    "course_name"
)["student_id"].nunique().sort_values(ascending=False)
unique_students
```

**Q10. students aur courses ko merge karne ke bad har student ke courses ki list nikalo.**

**Answer:**

```python
students_courses
```

```python
student_course_list=students_courses.groupby(
    ["student_id","name"]
)[["course_name","course_id"]].agg(list).reset_index()
student_course_list
```
