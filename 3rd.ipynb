# PROGRAM 4:USER-DEFINED FUNCTION+FILTRING+VALUE_COUNTS()|

```python
def get_city_matches(city):
    result=data[data["city"]==city]
    return result
```

```python
hyderabad_matches=get_city_matches("Indore")
```

```python
hyderabad_matches
```

```python
print("Hyderabad matches:",hyderabad_matches.shape)
```

```python
def get_team_wins(team_name):
    result=data[data["winner"]==team_name]
    return result[["season","city","team1","team2","winner"]]
```

```python
get_team_wins("Sunrisers Hyderabad")
```

```python
winner_count=data["winner"].value_counts()
```

```python
print(winner_count)
```

```python
print("Most frequent winner:",winner_count.index[0])
```
