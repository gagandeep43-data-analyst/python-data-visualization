# 1. Relational Plot

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
# scatter plot->axes level function
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# relplot-> figure level->square shape
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
# scatter using relplot-> size and hue
```

```python
# style semantic
```

```python
# line plot
gap=px.data.gapminder()
team_df=gap[gap['country']=='India']
team_df
```

```python
# axes level function
```

```python
sns.lineplot(data=team_df,x='year',y='lifeExp')
```

```python
# using relplot
```

```python
sns.relplot(data=team_df,x='year',y='lifeExp',kind='line')
```

```python
# hue-> style
```

```python
team_df=gap[gap['country'].isin(['India','Brazil','Germany'])]
team_df
```

```python
sns.lineplot(data=team_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
sns.relplot(kind='line',data=team_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
sns.lineplot(data=team_df,x='year',y='lifeExp',hue='country')
```

```python
# facet plot-> figure level function->work with relplot
# it will not work with scatterplot and lieplot
sns.relplot(data=tips,x='total_bill',y='tip',kind='line',col='sex',row='day')
```

```python
# col wrap
```

```python
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=4)
```

```python
sns.relplot(kind='scatter',data=tips,x='total_bill',y='tip',col='sex',row='day')
```

```python
# figure level-> displot
# axes level->histplot->kdeplot->rugplot
```

```python
tips
```

```python
# plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=2)
```

```python
# it's also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discreate bins are automatically set for cate

# countplot
sns.displot(data=tips,x='day',kind='hist')
```

```python
# hue parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
# element-> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',hue='sex')
```

```python
# faceting using col and row-> not work on histplot function
```

```python
sns.displot(data=tips,x='tip',kind='hist',element='step',hue='sex')
```

```python
#kdeplot
# Rather than using distrete bins, a KDE plot smooth the observation with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
# hue-> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=20,aspect=4)
```

```python
# Rugplot
```

```python
# Plot marginal distributions by drawing ticks along the x and y axes.
#This function is intended to complement other plots by showing the locations of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within ractangles that tile the plor
# and then shows the count of observations within each rectangle with the fill color
# sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
sns.displot(data=tips,x='total_bill',y='tip',kind='hist')
```

```python
# bivariate kdeplot
```

```python
# a bivariate kde plot smoothes the(x,y)observations with a 20 gaussian
```

```python
sns.kdeplot(data=tips,x='total_bill',y='tip')
```
