```python
marks=[67,78,90,88]
```

```python
subjects=['hindi','maths','science','sst']
```

```python
s4=pd.Series(marks,index=subjects)
```

```python
s4
```

```python
marks_dict={'math':67,'hindi':78,'english':90,'science':87}
```

```python
marks_Series=pd.Series(marks_dict,name='s1 marks')
```

```python
print(marks_Series.dtype)
```

```python
print(marks_Series.size)
```

```python
print(marks_Series.name)
```

```python
print(marks_Series.is_unique)
```

```python
print(marks_Series.index)
```
