# Program 5:Top 10 Teams by Wins-Line Chart

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("matches - matches (1).csv.xls")
data
```

```python
team_wins=data["winner"].value_counts()
team_wins
```

```python
top_teams=team_wins.sort_values(ascending=False).head(10)
top_teams
```

```python
top_teams.plot(kind="line")
plt.xlabel("Team")
plt.ylabel("Number of Wins")
plt.title("Top 10 Teams by Number of wins")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

# Program 6:Toss Decision-Scatter Chart

```python
data
```

```python
toss_count=data["toss_decision"].value_counts()
toss_count
```

```python
toss_count=toss_count.sort_values(ascending=False)
toss_count
```

```python
x=range(len(toss_count))
y=toss_count.values
plt.scatter(x,y,s=120)
plt.xticks(x,toss_count.index)
plt.ylabel("Toss Decision")
plt.title("Toss Decision Distribution")
plt.grid(True)
plt.show
```
