```python
sales["Discount"]=sales["Discount"].fillna(sales["Discount"].mean())
sales
```

```python
sales["Quantity"]=sales["Quantity"].astype("int64")
sales
```

```python
sales["Discount"]=sales["Discount"].astype("float64")
sales
```

```python
sales["Gross_Sales"]=sales["Quantity"]*sales["Price"]
sales
```

```python
sales["Discount_Amount"]=sales["Gross_Sales"]*sales["Discount"]/100
sales
```

```python
sales["Net_Sales"]=sales["Gross_Sales"]-sales["Discount_Amount"]
sales
```

```python
sales["Net_Sales"]=sales["Net_Sales"].clip(lower=0)
sales["Net_Sales"]
```

```python
sales["Order_Type"]=sales["Net_Sales"].apply(lambda x: "High Value" if x >=80000 else "Medium Value" if x >=40000 else"Low Value")
sales
    
```

```python
extra_orders=pd.DataFrame({
    "Order_ID":[1013,1014],
    "Customer":["Guri","Harry"],
    "Product":["Laptop","Phone"],
    "Category":["Electronics","Accessories"],
    "Quantity":[1,3],
    "Price":[55000,8000],
    "Discount":[7,5],
    "City":["Delhi","Sirsa"]
})
extra_orders
```

```python
sales_extended=pd.concat([sales,extra_orders],ignore_index=True)
sales_extended
```

```python
City_data=pd.DataFrame({
    "City":["Delhi","Sirsa","Jaipur","Hisar"],
    "Zone":["North","North","West","North"],
    "State":["Delhi","Haryana","Rajasthan","Haryana"]
})
City_data
```

```python
sales_final=pd.merge(sales_extended,City_data,on="City",how="left")
sales_final
```

```python
Product_counts=sales_final["Product"].value_counts()
Product_counts
```

```python
Category_counts=sales_final["Category"].value_counts()
Category_counts
```

```python

plt.figure(figsize=(9,5))
plt.bar(Product_counts.index,Product_counts.values)
plt.xlabel("Product")
plt.ylabel("Net Sales")
plt.title("Product Wise Net Sales")
plt.xticks(rotation=45)
plt.tight_layout
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(Category_counts.values,labels=Category_counts.index)
plt.title("Category Distribution")
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.plot(sales_final["Oder_ID"],sales_final["Net_Sales"])
plt.xlabel("Order_ID")
plt.ylabel("Net_Sales")
plt.title("Order Wise net sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```
