```python
bonus=pd.DataFrame({
    "Student_ID":[111,112],
    "Name":["Kavir","Tanya"],
    "Age":[21,20],
    "Marks":[82,90],
    "Attendance":[87,94],
    "Department":["AI","Python"],
    "Fees":[43500,45000],
    "Grade":["B","A"]
})
bonus
```

```python
students_extended=pd.concat([students,bonus],ignore_index=True)
students_extended
```

```python
department_data=pd.DataFrame({
    "Department":["Data Science","AI","Python"],
    "Mentor":["Anita","Rohit","Kunal"],
    "Room":[101,202,303]
})
department_data
```

```python
students_final=pd.merge(students_extended,department_data,on="Department",how="left")
students_final
```

```python
department_average=students_final.groupby("Department")["Marks"].mean()
department_average
```
