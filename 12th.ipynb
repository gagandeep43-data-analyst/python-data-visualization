# Pandas Merge-Program 1

```python
import pandas as pd
```

```python
students=pd.read_csv("students.csv.xls")
students.tail(2)
```

```python
courses=pd.read_csv("courses.csv.xls")
courses.head(2)
```

**Q1.Students aur courses ko course_id ke basis par merge registration details ka Dataframe banao.**

**Answer:**

```python
reg=pd.read_csv("reg-month1.csv.xls")
result=pd.merge(reg,students, on="student_id", how="inner")
result
result=pd.merge(result,courses, on="course_id", how="inner")
result
```

**Q2. students aur month 1 registrations ki inner merge karo**

**Answer:**

```python
students.head()
```

```python
reg.head()
```

```python
inner_result=pd.merge(
    students,
    reg,
    on="student_id",
    how="inner"
)
inner_result
```

**Q3. students aur month 1 registrations ko left merge karo.**

**Answer:**

```python
left_result=pd.merge(
    students,
    reg,
    on="student_id",
    how="left"
)
left_result
```

**Q4. Month 1 registration aur courses ko course_id par merge karke course name aur price add karo.**

**Answer:**

```python
reg.head()
```

```python
courses.head()
```

```python
course_result=pd.merge(
    reg,
    courses,
    on="course_id",
    how="inner"
)
course_result
A=pd.merge(course_result,students,on='student_id',how='inner')
A
```

**Q5. students ke saath courses ki complete information add karke final table banao.**

**Answer:**


```python
final_result=pd.merge(
    inner_result,
    courses,
    on="course_id",
    how="left"
)
final_result
```

**Q6.har course ki registration count aur total revenue nikalo.**

**Answer:**

```python
summary=final_result.groupby("course_name").agg(
    registrations=("student_id","count"),
    total_revenue=("price","sum")
).sort_values("registrations",ascending=False)
summary
```

```python
summary=final_result.groupby("course_name").agg(
       registrations=("student_id","count"))
summary
```
