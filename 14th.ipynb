# Pandas Merge - Program 3

```python
import pandas as pd
```

```python
students=pd.read_csv("students.csv.xls")
```

```python
month1=pd.read_csv("reg-month1.csv.xls")
```

```python
month2=pd.read_csv("reg-month2.csv.xls")
```

```python
courses=pd.read_csv("courses.csv.xls")
```

**Q1.Month 1 aur Month 2 regisraton data ko merge karke ek complete registration table bnao.**

**Answer:**

```python
registrations=pd.concat(
    [
        month1.assign(month="Month 1"),
        month2.assign(month="Month 2")
   ],
    ignore_index=True
  )  
data=pd.merge(
        registrations,
        students,
        on="student_id",
        how="left"
    )
data=pd.merge(
       data,
       courses,
       on="course_id",
       how="left"
   )
data
```

**Q2. Har registration ke saath student ka naam,course name aur course price dikhao.**

**Answer:**

```python
data[
    [
        "student_id",
        "name",
        "course_name",
        "price",
        "month"
    ]
]
```

**Q3. Har course ke liye Month 1 aur Month 2 ki registrations count karo.**

**Answer:**

```python
course_month=data.groupby(
    ["course_name","month"]
).size().reset_index(name="registration")
course_month
```

**Q4. har course ki total registrations, unique students aur total revenue nikalo.**

**Answer:**

```python
course_summary=data.groupby("course_name").agg(
    total_registrations=("student_id","count"),
    unique_students=("student_id","nunique"),
    total_revenue=("price","sum")
).sort_values(
    "total_registrations",
    ascending=False
)
course_summary
```

**Q5.Har registration ke sath us course ki total registration count add karo.**

**Answer:**

```python
data["course_total_registrations"]=data.groupby(
    "course_id"
)["student_id"].transform("count")
print(
data[
    [
        "student_id",
        "name",
        "course_name",
        "month",
        "course_total_registrations"
    ]
 ]
)
data

        
        
```
