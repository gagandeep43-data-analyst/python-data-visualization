# Pandas Merge-Program 2

```python
import pandas as pd
```

```python
students=pd.read_csv("students.csv.xls")
```

```python
month1=pd.read_csv("reg-month1.csv.xls")
```

```python
month2=pd.read_csv("reg-month2.csv.xls")
```

```python
courses=pd.read_csv("courses.csv.xls")
```

**Q1. Month 1 aur Month 2 registrations ko students dadaset ke sasth merge karke dono ka complete student data banao.**

**Answer:**

```python
students.head()
```

```python
month1.head()
```

```python
month2.head()
```

```python
courses.head()
```

```python
m1=pd.merge(
    month1,
    students,
    on="student_id",
    how="inner"
)
m1
```

```python
m2=pd.merge(
    month2,
    students,
    on="student_id",
    how="inner"
)
m2
```

```python
m2=pd.merge(
    m1,
    month2,
    on="student_id",
    how="inner"
)
m2
```

```python
courses
```

**Q2. Month 2 ke data ko courses ke sath merge karke course ka name aur price add karo.**

**Answer:**

```python
m2_course=pd.merge(
    month2,
    courses,
    on="course_id",
    how="inner"
)
m2_course
```

**Q3. month 1 ke data ko courses ke sath merge karke course ka name aur price add karo.**

**Answer:**

```python
m1_course=pd.merge(
     month1,
    courses,
    on="course_id",
    how="inner"
)
m1_course
```

**Q4. dono months ke merged data ko ek single DataFrame me combine karo.**

**Answer:**

```python
final=pd.concat(
    [m1_course,m2_course],
    ignore_index=True
)
final
```

**Q5. Har student ne total kitne courses liye aue total kitna spend kiya, ye nikalo.**

**Answer:**

```python
student_summary=final.groupby(
    ["student_id"]
).agg(
total_courses=("course_id","count"),
total_spent=("price","sum")
).sort_values("total_courses",ascending=False)
student_summary
```

**Q6. Sabse zyada courses lene wale student ki complete information nikalo.**

**Answer:**

```python
top_student=student_summary.loc[
    student_summary["total_courses"].idxmax()
    ]
top_student
```

**Answer:**
