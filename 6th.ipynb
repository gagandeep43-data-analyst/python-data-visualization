# program 2: Top 5 Match Winners with Bar Chart

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
data
```

```python
Winner_count =data["TossWinner"].value_counts()
Winner_count
```

```python
Top5=Winner_count.head(5)
Top5
```

```python
Top5.plot(kind="bar")
plt.xlabel("Team")
plt.ylabel("Number of wins")
plt.title("Top 5 Teams by Number of Wins")
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```
