# Pandas GroupBy - Big Programe 2

```python
import pandas as pd
```

```python
df=pd.read_csv("imdb-top-1000.csv.xls")
df
```

```python
df["Released_Year"]=pd.to_numeric(df["Released_Year"],errors="coerce")
df
```

```python
df["IMDB_Rating"]=pd.to_numeric(df["IMDB_Rating"],errors="coerce")
df
```

```python
df["No_of_Votes	"]=pd.to_numeric(df["No_of_Votes"],errors="coerce")
df
```

```python
df["Gross"]=pd.to_numeric(df["Gross"],errors="coerce")
df
```

```python
df["Metascore"]=pd.to_numeric(df["Metascore"],errors="coerce")
df
```

```python
director_rating=df.groupby("Director")["IMDB_Rating"].mean().sort_values(ascending=False)
director_rating
```

```python
director_analysis=df.groupby("Director",as_index=False).agg(
     Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","max"),
    Highest_Rating=("IMDB_Rating","mean"),
    Average_Runtime=("Runtime","mean"),
    Total_Votes=("No_of_Votes","sum"),
    Total_Gross=("Gross","sum")
    )
director_analysis
    
```

```python
director_analysis=director_analysis.sort_values(
    ["Movies","Average_Rating"],
    ascending=[False,False]
)
director_analysis
```

```python
director_analysis.head(20)
```

```python
experienced_directors=director_analysis[
    director_analysis["Movies"] >=3
    ].sort_values("Average_Rating",ascending=False)
experienced_directors.head(20)
```

```python
top_director=director_analysis[(director_analysis["Movies"] >=3)&(director_analysis["Average_Rating"] >=7.5)
    ]
top_director
```

```python
top_director.sort_values("Average_Rating",ascending=False)
```

```python
best_director= director_analysis.loc[director_analysis["Average_Rating"].idxmax()]
best_director
```

```python
most_movies_director=director_analysis.loc[director_analysis["Movies"].idxmax()]
most_movies_director
```

```python
highest_gross_director=director_analysis.loc[director_analysis["Total_Gross"].idxmax()]
highest_gross_director
```

```python
director_groups=df.groupby("Director")
```

```python
director_groups.get_group("Christopher Nolan")
```

```python
director_filter= director_groups.filter(lambda x: len(x) >=3)
director_filter
```

```python
director_mean_rating=director_groups["IMDB_Rating"].transform("mean")
```

```python
df["Director_Average_Rating"]=director_mean_rating
```

```python
print(
    df[[
        "Series_Title",
        "Director",
        "IMDB_Rating",
        "Director_Average_Rating"
    ]].sort_values(
        "Director_Average_Rating",
        ascending=False
    ).head(20)
)
```

```python
director_genre= df.groupby(
    ["Director","Genre"],
    as_index=False
).agg(
    Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","mean"),
    Total_Votes=("No_of_Votes","sum")
)
director_genre=director_genre.sort_values(
    ["Director","Average_Rating"],
    ascending=[True,False]
               )
director_genre.head(30)
```

```python
star_analysis=df.groupby("Star1",as_index=False).agg(
     Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","mean"),
    Highest_Rating=("IMDB_Rating","max"),
    Total_Votes=("No_of_Votes","sum") 
    )
star_analysis=star_analysis[
       star_analysis["Movies"] >=3
       ].sort_values(
       "Average_Rating",
 ascending=False
 )
star_analysis.head(20)
    
```

```python
director_status=df.groupby("Director").agg(
     Movies=("Series_Title","count"),
    Average_Rating=("IMDB_Rating","mean")
)
director_status["Rating_Rank"]=director_status["Average_Rating"].rank(ascending=False,method="dense")
director_status.sort_values("Rating_Rank").head(20)
```
